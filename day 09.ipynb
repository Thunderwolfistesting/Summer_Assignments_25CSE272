33. Write a program to Print reverse star pattern.
```text
*****
****
***
**
*
```

````python
size = 5

for i in range(size, 0, -1): # loop downwards
    for j in range(i): # row generation
        print("*", end="")
    print()
````

34. Write a program to Print reverse number triangle.
```text
12345
1234
123
12
1
```

````python
total_rows = 5

for r in range(total_rows, 0, -1): # count backwards
    for c in range(1, r + 1):
        print(c, end="") # output character numbers
    print()
````

35. Write a program to Print repeated character pattern.
```text
A
BB
CCC
DDDD
EEEEE
```

````python
limit = 5

for i in range(1, limit + 1):
    for j in range(i): # handle repeat counts
        print(chr(64 + i), end="") // ascii tracking
    print()
````

36. Write a program to Print hollow square pattern.
```text
*****
* *
* *
* *
*****
```

````python
side = 5

for idx in range(side):
    if idx == 0 or idx == side - 1: # top and bottom borders
        print("*" * side)
    else:
        print("*" + " " * (side - 2) + "*") # central gaps
````
