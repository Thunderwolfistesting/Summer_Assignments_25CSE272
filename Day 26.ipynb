101. Write a program to Create number guessing game.

```python
import random

hidden_target = random.randint(1, 100)

while True:
    player_guess = int(input("Enter your guess (1-100): "))
    print(f"You entered: {player_guess}")

    if player_guess < hidden_target:
        print("Too low! Try again.")
    elif player_guess > hidden_target:
        print("Too high! Try again.")
    else:
        print(f"Congratulations! You guessed the number {hidden_target}.")
        break
```

102. Write a program to Create voting eligibility system.

```python
voter_name = input("Enter your name: ")
voter_age = int(input("Enter your age: "))

print(f"Name: {voter_name}")
print(f"Age: {voter_age}")

if voter_age >= 18:
    print(f"{voter_name}, you are eligible to vote.")
else:
    print(f"{voter_name}, you are not eligible to vote.")
```

103. Write a program to Create ATM simulation.

```python
account_balance = 2500

while True:
    print("\n-- ATM MENU --")
    print("1. Check Balance")
    print("2. Deposit")
    print("3. Withdraw")
    print("4. Exit")

    user_selection = int(input("Enter your choice: "))

    if user_selection == 1:
        print(f"Your current balance is: {account_balance}")

    elif user_selection == 2:
        deposit_funds = float(input("Enter deposit amount: "))
        account_balance += deposit_funds
        print(f"You deposited: {deposit_funds}")
        print(f"New balance: {account_balance}")

    elif user_selection == 3:
        withdrawal_funds = float(input("Enter withdrawal amount: "))
        if withdrawal_funds <= account_balance:
            account_balance -= withdrawal_funds
            print(f"You withdrew: {withdrawal_funds}")
            print(f"Remaining balance: {account_balance}")
        else:
            print("Insufficient balance!")

    elif user_selection == 4:
        print("Thank you for using the ATM!")
        break

    else:
        print("Invalid choice! Please try again.")
```

104. Write a program to Create quiz application.

```python
running_score = 0

print("-- Tech Quiz --")

quiz_reply = input("1. Which programming language is known as the language of AI and Machine Learning? ")
print(f"You entered: {quiz_reply}")

if quiz_reply.lower() == "python":
    print("Correct!")
    running_score += 1
else:
    print("Wrong! The correct answer is Python.")

quiz_reply = input("2. What does CPU stand for? ")
print(f"You entered: {quiz_reply}")

if quiz_reply.lower() == "central processing unit":
    print("Correct!")
    running_score += 1
else:
    print("Wrong! The correct answer is Central Processing Unit.")

quiz_reply = input("3. Which company developed the Windows operating system? ")
print(f"You entered: {quiz_reply}")

if quiz_reply.lower() == "microsoft":
    print("Correct!")
    running_score += 1
else:
    print("Wrong! The correct answer is Microsoft.")

quiz_reply = input("4. Which data structure works on the LIFO principle? ")
print(f"You entered: {quiz_reply}")

if quiz_reply.lower() == "stack":
    print("Correct!")
    running_score += 1
else:
    print("Wrong! The correct answer is Stack.")

quiz_reply = input("5. What does HTML stand for? ")
print(f"You entered: {quiz_reply}")

if quiz_reply.lower() == "hypertext markup language":
    print("Correct!")
    running_score += 1
else:
    print("Wrong! The correct answer is HyperText Markup Language.")

print(f"\nYour final score is: {running_score}/5")
```
