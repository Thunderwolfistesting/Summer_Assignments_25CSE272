13. Write a program to Generate Fibonacci series.

```python
terms = int(input("Enter total terms: "))
f1, f2 = 0, 1

print(f"Fibonacci series for {terms}:")
for _ in range(terms):
    print(f1, end=" ")
    f1, f2 = f2, f1 + f2
```

14. Write a program to Find nth Fibonacci term

```python
pos = int(input("Enter position: "))
x, y = 0, 1

for _ in range(pos - 1):
    x, y = y, x + y

print(f"The term at {pos}th is {x}")
```

15. Write a program to Check Armstrong number.

```python
val = int(input("Enter value: "))
pwr = len(str(val)) # find number of digits
t = val
res = 0

while t > 0:
    rem = t % 10
    res += rem ** pwr
    t //= 10

if res == val:
    print(f"The {val} is an armstrong number")
else:
    print(f"The {val} is not an armstrong number")
```

16. Write a program to Print Armstrong numbers in a range.

```python
low = int(input("Start: "))
high = int(input("End: "))

print(f" The Armstrong no. between {low} and {high} are: ")

for current in range(low, high + 1):
    chk = current
    digits = len(str(current))
    running_sum = 0

    while chk > 0:
        rem = chk % 10
        running_sum += rem ** digits
        chk //= 10

    if running_sum == current:
        print(current)
```
