17. Write a program to Check perfect number.

```python
n = int(input("Enter a number: "))
total = 0

for i in range(1, n):
    if n % i == 0: # check if it divides cleanly
        total += i

if total == n: # matches the original number
    print(n, "is a Perfect Number")
else:
    print(n, "is not a Perfect Number")
```

18. Write a program to Check strong number.

```python
check_val = int(input("Enter a number: "))
saved = check_val
fact_sum = 0

while check_val > 0:
    rem = check_val % 10 # isolate each digit
    f = 1
    for j in range(1, rem + 1):
        f *= j
    fact_sum += f
    check_val //= 10 # move to next digit

if fact_sum == saved:
    print(saved, "is a Strong Number")
else:
    print(saved, "is not a Strong Number")
```

19. Write a program to Print factors of a number.

```python
item = int(input("Enter a number: "))
print(f"Factors of {item} are:") # header for output list

for i in range(1, item + 1):
    if item % i == 0: # found a factor
        print(i, end=" ")
```

20. Write a program to Find largest prime factor.

```python
target = int(input("Enter a number: "))
factor_limit = 0
div = 2

while div * div <= target: # loop through possible factr
    while target % div == 0:
        factor_limit = div
        target //= div
    div += 1 # test next value

if target > 1:
    factor_limit = target

print("Largest Prime Factor =", factor_limit)
```
