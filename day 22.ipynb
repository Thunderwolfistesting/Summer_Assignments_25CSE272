85. Write a program to Check palindrome string.

```python
input_data = input("Enter a string: ")

reversed_data = ""

for letter in input_data:
    reversed_data = letter + reversed_data # accumulate backward string slice

print(f"You entered: {input_data}")

if input_data == reversed_data:
    print("The string is a palindrome.")
else:
    print("The string is not a palindrome.")
```

86. Write a program to Count words in a sentence.

```python
input_data = input("Enter a sentence: ")

word_tally = 1

for letter in input_data:
    if letter == ' ':
        word_tally += 1 # increment word intervals

print(f"You entered: {input_data}")
print(f"Number of words: {word_tally}")
```

87. Write a program to Character frequency.


```python
input_data = input("Enter a string: ")
target_char = input("Enter the character to find: ")

match_counter = 0

for current_char in input_data:
    if current_char == target_char:
        match_counter += 1

print(f"You entered: {input_data}")
print(f"Frequency of '{target_char}' is: {match_counter}")
```

88. Write a program to Remove spaces from a string.


```python
input_data = input("Enter a string: ")

compact_output = ""

for letter in input_data:
    if letter != ' ':
        compact_output = compact_output + letter # filter space instances

print(f"You entered: {input_data}")
print(f"String without spaces: {compact_output}")
```
