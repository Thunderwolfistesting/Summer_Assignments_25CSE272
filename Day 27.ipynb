105. Write a program to Create student record management system.

```python
student_database = []

while True:
    print("\n-- Student Record Management System ---")
    print("1. Add Student")
    print("2. View Students")
    print("3. Exit")

    menu_selection = int(input("Enter your choice: "))

    if menu_selection == 1:
        full_name = input("Enter student name: ")
        id_number = input("Enter roll number: ")
        grade_score = float(input("Enter marks: "))

        student_database.append([full_name, id_number, grade_score])
        print(f"Student '{full_name}' added successfully!")

    elif menu_selection == 2:
        if len(student_database) == 0:
            print("No student records found.")
        else:
            print("\n----- Student Records -----")
            for record in student_database:
                print(f"Name: {record[0]}")
                print(f"Roll No: {record[1]}")
                print(f"Marks: {record[2]}")
                print("---------------------------")

    elif menu_selection == 3:
        print("Exiting Student Record Management System.")
        break
    else:
        print("Invalid choice! Please try again.")
```

106. Write a program to Create employee management system.


```python
employee_roster = []

while True:
    print("\n-- Employee Management System ---")
    print("1. Add Employee")
    print("2. View Employees")
    print("3. Exit")

    menu_selection = int(input("Enter your choice: "))

    if menu_selection == 1:
        staff_id = input("Enter Employee ID: ")
        full_name = input("Enter Employee Name: ")
        dept_name = input("Enter Department: ")
        base_pay = float(input("Enter Salary: "))

        employee_roster.append([staff_id, full_name, dept_name, base_pay])
        print(f"Employee '{full_name}' added successfully!")

    elif menu_selection == 2:
        if len(employee_roster) == 0:
            print("No employee records found.")
        else:
            print("\n----- Employee Records -----")
            for staff in employee_roster:
                print(f"Employee ID: {staff[0]}")
                print(f"Name: {staff[1]}")
                print(f"Department: {staff[2]}")
                print(f"Salary: {staff[3]}")
                print("----------------------------")

    elif menu_selection == 3:
        print("Exiting Employee Management System.")
        break
    else:
        print("Invalid choice! Please try again.")
```

107. Write a program to Create salary management system.

```python
payroll_database = []

while True:
    print("\n--- Salary Management System ---")
    print("1. Add Employee Salary")
    print("2. View Employee Salaries")
    print("3. Exit")

    menu_selection = int(input("Enter your choice: "))

    if menu_selection == 1:
        staff_id = input("Enter Employee ID: ")
        full_name = input("Enter Employee Name: ")
        base_pay = float(input("Enter Salary: "))

        payroll_database.append([staff_id, full_name, base_pay])
        print(f"Salary record for {full_name} added successfully!")

    elif menu_selection == 2:
        if len(payroll_database) == 0:
            print("No salary records found.")
        else:
            print("\n----- Salary Records -----")
            for entry in payroll_database:
                print(f"Employee ID : {entry[0]}")
                print(f"Name        : {entry[1]}")
                print(f"Salary      : {entry[2]}")
                print("--------------------------")

    elif menu_selection == 3:
        print("Exiting Salary Management System.")
        break
    else:
        print("Invalid choice! Please try again.")
```

108. Write a program to Create marksheet generation system.


```python
student_name = input("Enter student name: ")
roll_number = input("Enter roll number: ")

math_score = float(input("Enter Math marks: "))
science_score = float(input("Enter Science marks: "))
english_score = float(input("Enter English marks: "))

combined_total = math_score + science_score + english_score
average_percentage = combined_total / 3

if average_percentage >= 90:
    final_grade = "A+"
elif average_percentage >= 80:
    final_grade = "A"
elif average_percentage >= 70:
    final_grade = "B"
elif average_percentage >= 60:
    final_grade = "C"
elif average_percentage >= 50:
    final_grade = "D"
else:
    final_grade = "F"

print("\n===== Student Marksheet =====")
print(f"Name       : {student_name}")
print(f"Roll No    : {roll_number}")
print(f"Math       : {math_score}")
print(f"Science    : {science_score}")
print(f"English    : {english_score}")
print(f"Total      : {combined_total}")
print(f"Percentage : {average_percentage:.2f}%")
print(f"Grade      : {final_grade}")

if final_grade == "F":
    print("Result     : Fail")
else:
    print("Result     : Pass")
```
