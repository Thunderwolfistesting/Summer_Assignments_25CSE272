81. Write a program to Find string length without strlen().

```python
user_str = input("Enter a string: ")

length_tally = 0

for character in user_str:
    length_tally += 1
print(f"you enter {user_str}")
print(f"The length of the string is: {length_tally}")
```

82. Write a program to Reverse a string.

```python
user_str = input("Enter a string: ")

inverted_str = ""

for character in user_str: # prepend characters to invert
    inverted_str = character + inverted_str

print(f"You entered: {user_str}")
print(f"Reversed string: {inverted_str}")
```

83. Write a program to Count vowels and consonants.

```python
user_str = input("Enter a string: ")

v_count = 0
c_count = 0

for character in user_str:
    if ('A' <= character <= 'Z') or ('a' <= character <= 'z'):
        if character == 'A' or character == 'E' or character == 'I' or character == 'O' or character == 'U' or \
           character == 'a' or character == 'e' or character == 'i' or character == 'o' or character == 'u':
            v_count += 1
        else: # tracking consonants separately
            c_count += 1

print(f"You entered: {user_str}")
print(f"Number of vowels: {v_count}")
print(f"Number of consonants: {c_count}")
```

84. Write a program to Convert lowercase to uppercase.

```python
user_str = input("Enter a string: ")

upper_output = ""

for character in user_str:
    if character >= 'a' and character <= 'z':
        upper_output = upper_output + chr(ord(character) - 32) # perform ascii shifting
    else:
        upper_output = upper_output + character

print(f"You entered: {user_str}")
print(f"Uppercase: {upper_output}")
```
