65. Write a program to Merge arrays.

```python
def combine_sorted_lists(first_list, second_list):
    combined = []
    p1, p2 = 0, 0

    while p1 < len(first_list) and p2 < len(second_list):
        if first_list[p1] <= second_list[p2]:
            combined.append(first_list[p1])
            p1 += 1
        else:
            combined.append(second_list[p2])
            p2 += 1
        
    while p1 < len(first_list): # push left overs
        combined.append(first_list[p1])
        p1 += 1

    while p2 < len(second_list): # push right overs
        combined.append(second_list[p2])
        p2 += 1
    return combined

left_side = [2, 4, 7, 9]
right_side = [3, 6, 8, 10]

print(combine_sorted_lists(left_side, right_side))
```

66. Write a program to Union of arrays.

```python
dataset1 = list(map(int, input("Enter first array elements: ").split()))
dataset2 = list(map(int, input("Enter second array elements: ").split()))

distinct_union = []

for item in dataset1:
    if item not in distinct_union: # filter unique elements
        distinct_union.append(item)

for item in dataset2:
    if item not in distinct_union: # evaluate second loop tracking
        distinct_union.append(item)

print(f"First Array: {dataset1}")
print(f"Second Array: {dataset2}")
print(f"Union of Arrays: {distinct_union}")
```

67. Write a program to Intersection of arrays.

```python
sequence1 = list(map(int, input("Enter first array elements: ").split()))
sequence2 = list(map(int, input("Enter second array elements: ").split()))

matched_intersection = []

for element in sequence1:
    if element in sequence2 and element not in matched_intersection: # logical check match
        matched_intersection.append(element)

print(f"First Array: {sequence1}")
print(f"Second Array: {sequence2}")
print(f"Intersection of Arrays: {matched_intersection}")
```

68. Write a program to Find common elements.

```python
group_a = list(map(int, input("Enter first array elements: ").split()))
group_b = list(map(int, input("Enter second array elements: ").split()))

shared_items = []

for value in group_a:
    if value in group_b and value not in shared_items: # find duplicate targets
        shared_items.append(value)

print(f"First Array: {group_a}")
print(f"Second Array: {group_b}")
print(f"Common Elements: {shared_items}")
```
