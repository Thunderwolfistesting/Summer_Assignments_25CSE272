57. Write a program to Reverse array.

```python
items = [1, 3, 5, 7, 9]
inverted_list = []

while len(items) > 0: # run element extract step
    inverted_list.append(items.pop())

print(f"Reversed array:{inverted_list}")
```

```python
collection = [1, 3, 5, 7, 9]
length = len(collection)

for idx in range(length // 2): # run inversion logic
    collection[idx], collection[length - 1 - idx] = collection[length - 1 - idx], collection[idx]

print(f"Reversed array:{collection}")
```

58. Write a program to Rotate array left.

```python
values = [10, 20, 30, 40, 50]
shift_steps = 2

normalized_shift = shift_steps % len(values) # wrap bounds calc
left_rotated = values[normalized_shift:] + values[:normalized_shift]

print(f"Left rotated array:{left_rotated}")
```

59. Write a program to Rotate array right.

```python
dataset = [10, 20, 30, 40, 50]
offset = 2

split_point = len(dataset) - (offset % len(dataset)) # track split index
right_rotated = dataset[split_point:] + dataset[:split_point]

print(f"Right rotated array:{right_rotated}")
```

60. Write a program to Move zeroes to end.

```python
numbers = [0, 4, 0, 5]

numbers.sort(key=lambda x: x == 0) # group values by condition

print(f"Array after moving zeroes to end:{numbers}")
```

```python
numbers_list = [0, 4, 0, 5]
target_idx = 0

for i in range(len(numbers_list)):
    if numbers_list[i] != 0:
        numbers_list[target_idx] = numbers_list[i] # overwrite active slots
        if i != target_idx:
            numbers_list[i] = 0
        target_idx += 1

print(f"Array after moving zeroes to end:{numbers_list}")
```
