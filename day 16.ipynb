61. Write a program to Find missing number in array.


```python
def locate_missing(vals, limit_val):
    expected_sum = limit_val * (limit_val + 1) // 2

    current_total = 0
    for element in vals: # sum elements
        current_total += element

    return expected_sum - current_total

sample_list = [1, 2, 3, 5, 6, 7]
max_n = 7
missing_num = locate_missing(sample_list, max_n) # evaluate delta
print(f"Missing Number: {missing_num}")
```

62. Write a program to Find maximum frequency element.

```python
def highest_freq_item(items):
    top_count = 0
    mode_element = items[0]

    for idx in range(len(items)):
        current_occurrences = 0

        for inner_idx in range(len(items)): # nested verification pass
            if items[idx] == items[inner_idx]:
                current_occurrences += 1

        if current_occurrences > top_count:
            top_count = current_occurrences
            mode_element = items[idx] # preserve target metadata

    return mode_element, top_count

dataset = [8, 2, 8, 4, 5, 8, 9]

matched_item, total_hits = highest_freq_item(dataset)
print(f"Maximum Frequency Element: {matched_item}")
print(f"Frequency: {total_hits}")
```

63. Write a program to Find pair with given sum.

```python
def match_sum_pair(values_list, sum_goal):
    for i in range(len(values_list)):
        for j in range(i + 1, len(values_list)): # verify upcoming indices
            if values_list[i] + values_list[j] == sum_goal:
                return values_list[i], values_list[j]

    return None # edge case protection

numbers_set = [2, 11, 15, 6, 7]
target_val = 8

coords = match_sum_pair(numbers_set, target_val)

if coords:
    print(f"Pair found: {coords[0]} and {coords[1]}")
else:
    print(f"No pair found with sum {target_val}")
```

64. Write a program to Remove duplicates from array.

```python
def strip_repeats(collection):
    cleared_list = []

    for value in collection:
        if value not in cleared_list: # uniqueness check
            cleared_list.append(value)

    return cleared_list # output array mapping

raw_data = [10, 20, 20, 30, 10, 30]
print(f"Array after removing duplicates: {strip_repeats(raw_data)}")
```
