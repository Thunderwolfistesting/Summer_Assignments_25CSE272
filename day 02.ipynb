5. Write a program to Find sum of digits of a number.

```python
n = int(input("Enter number: "))
orig = n
digit_sum = 0

while n > 0:
    rem = n % 10 # get last digit
    digit_sum += rem
    n = n // 10 # remove last digit

print(f"The sum of digits of a {orig} is {digit_sum}")
```

6. Write a program to Reverse a number.

```python
val = int(input("Enter number: "))
backup = val
reverse_num = 0

while val != 0:
    last_digit = val % 10
    reverse_num = (reverse_num * 10) + last_digit # build reversed number
    val = val // 10

print(f"The reverse of of a number {backup} is {reverse_num}")
```

7. Write a program to Find product of digits.

```python
user_num = int(input("Enter number: "))
initial = user_num
total_prod = 1 # start at 1 for multiplication

while user_num > 0:
    rem = user_num % 10
    total_prod *= rem
    user_num //= 10

print(f"The product of digits of a {initial} is {total_prod}")
```

8. Write a program to Check whether a number is palindrome.

```python
x = int(input("Enter number: "))
original_x = x
inverted = 0

while x > 0:
    rem = x % 10
    inverted = (inverted * 10) + rem
    x //= 10

if inverted == original_x:
    print(f"{original_x} is Palindrome")
else:
    print(f"{original_x} is not a palindrome")
```
