93. Write a program to Check string rotation.

```python
s1 = input("Enter first string: ")
s2 = input("Enter second string: ")

print("First String: " + s1)
print("Second String: " + s2)

if len(s1) != len(s2):
    print("Strings are not rotations.")
else:
    temp = s1 + s1
    found = 0

    i = 0
    while i <= len(temp) - len(s2):
        j = 0

        while j < len(s2):
            if temp[i + j] != s2[j]:
                break
            j = j + 1

        if j == len(s2):
            found = 1
            break

        i = i + 1

    if found == 1:
        print("Strings are rotations.")
    else:
        print("Strings are not rotations.")
```

94. Write a program to Compress a string.

```python
s = input("Enter a string: ")

print("Input: " + s)

compressed = ""

i = 0

while i < len(s):
    count = 1

    while i + 1 < len(s) and s[i] == s[i + 1]:
        count = count + 1
        i = i + 1

    compressed = compressed + s[i] + str(count)
    i = i + 1

print("Compressed String: " + compressed)
```

95. Write a program to Find longest word.

```python
s = input("Enter a sentence: ")

print("Input: " + s)

maxWord = ""
word = ""

i = 0

while i < len(s):
    if s[i] != " ":
        word = word + s[i]
    else:
        if len(word) > len(maxWord):
            maxWord = word
        word = ""

    i = i + 1

if len(word) > len(maxWord):
    maxWord = word

print("Longest word: " + maxWord)
```

96. Write a program to remove duplicates

```python
s = input("Enter a string: ")

print("Input: " + s)

result = ""

i = 0

while i < len(s):
    found = 0
    j = 0

    while j < len(result):
        if s[i] == result[j]:
            found = 1
            break
        j = j + 1

    if found == 0:
        result = result + s[i]

    i = i + 1

print("String after removing duplicates: " + result)
```
