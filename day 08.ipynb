29. Write a program to Print half pyramid pattern.

````python
lines = int(input("Enter number of rows: "))

for i in range(1, lines + 1): # handle each row
    for j in range(i):
        print("*", end="")
    print() # line break
````

30. Write a program to Print number triangle.

```text
1
12
123
1234
```


````python
num_rows = int(input("Enter number of rows: "))

for r in range(1, num_rows + 1):
    for c in range(1, r + 1): # print increasing column values
        print(c, end="")
    print() # move to next line
````

31. Write a program to Print character triangle.
```text
A
AB
ABC
ABCD
```


````python
total_rows = int(input("Enter number of rows: "))

for i in range(total_rows):
    for j in range(i + 1): # loop up to current row index
        print(chr(65 + j), end="") # convert to character
    print()
````

32. Write a program to Print repeated-number pattern.
```text
1
22
333
4444
```


````python
limit = int(input("Enter number of rows: "))

for row_num in range(1, limit + 1):
    for col_num in range(row_num): # print item multiple times
        print(row_num, end="")
    print() # final row print
````
