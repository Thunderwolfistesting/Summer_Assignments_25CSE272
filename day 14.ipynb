53. Write a program to Linear search.

```python
def locate_item(items, target):
    idx = 0
    for entry in items:
        if entry == target: # scan current position
            return idx
        idx += 1
    return -1 # fallback flag

dataset = [15, 30, 45, 60, 75]
search_val = int(input("Enter the element to search: "))

match_idx = locate_item(dataset, search_val)

if match_idx != -1:
    print(f"Element found at index {match_idx}")
else:
    print("Element not found")
```

54. Write a program to Frequency of an element.

```python
total_inputs = int(input("Enter number of elements: "))
elements_list = []

for _ in range(total_inputs):
    elements_list.append(int(input("Enter element: ")))

print("Array elements are:")
for x in elements_list: # print current row configuration
    print(x, end=" ")

query_val = int(input("Enter the element to find frequency: "))
occurrence_count = 0

for item in elements_list:
    if item == query_val: # match counter logic
        occurrence_count += 1
        
print("")
print(f"Frequency of {query_val} is {occurrence_count}")
```

55. Write a program to Second largest element.

```python
def find_next_peak(numbers):
    highest = float("-inf")
    runner_up = float("-inf")
    
    for value in numbers: # parse values single pass
        if value > highest:
            runner_up = highest
            highest = value
        elif value > runner_up and value != highest: # mid range filter
            runner_up = value
            
    return runner_up

sample_set = [12, 45, 74, 5, 89, 23, 74]
print(find_next_peak(sample_set))
```

56. Write a program to Find duplicates in array.

```python
elements_count = int(input("Enter number of elements: "))
user_arr = []

for _ in range(elements_count):
    user_arr.append(int(input("Enter element: ")))
    
print("Array elements are:")
for entry in user_arr:
    print(entry, end=" ")

seen_elements = []
repeats = []

for element in user_arr: # check membership status
    if element in seen_elements:
        if element not in repeats: # guard duplicate listings
            repeats.append(element)
    else:
        seen_elements.append(element)

print("")
print(f"Duplicate elements are: {repeats}")
```
