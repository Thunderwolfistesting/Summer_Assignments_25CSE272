77. Write a program to Multiply matrices.

```python
r1 = int(input("Enter number of rows for Matrix 1: "))
c1 = int(input("Enter number of columns for Matrix 1: "))

r2 = int(input("\nEnter number of rows for Matrix 2: "))
c2 = int(input("Enter number of columns for Matrix 2: "))

if c1 != r2:
    print(f"\nMatrix multiplication is not possible because {c1} != {r2}.")
else:
    m1 = []
    print("\nEnter elements of Matrix 1:")
    for i in range(r1):
        row_vals = []
        for j in range(c1):
            row_vals.append(int(input(f"Enter element [{i}][{j}]: ")))
        m1.append(row_vals)

    m2 = []
    print("\nEnter elements of Matrix 2:")
    for i in range(r2):
        row_vals = []
        for j in range(c2):
            row_vals.append(int(input(f"Enter element [{i}][{j}]: ")))
        m2.append(row_vals)

    print("\nMatrix 1:")
    for single_row in m1:
        print(" ".join(f"{val}" for val in single_row))

    print("\nMatrix 2:")
    for single_row in m2:
        print(" ".join(f"{val}" for val in single_row))

    res_matrix = []
    for i in range(r1):
        zero_row = []
        for j in range(c2):
            zero_row.append(0)
        res_matrix.append(zero_row)

    for i in range(r1): # computation loops
        for j in range(c2):
            for k in range(c1):
                res_matrix[i][j] += m1[i][k] * m2[k][j]

    print("\nResultant Matrix:")
    for single_row in res_matrix: # render matrix structure
        print(" ".join(f"{val}" for val in single_row))
```

78. Write a program to Check symmetric matrix.

```python
dim_size = int(input("Enter the order of the square matrix: "))
grid = []

print("\nEnter the elements of the matrix:")
for r_idx in range(dim_size):
    row_buffer = []
    for c_idx in range(dim_size):
        row_buffer.append(int(input(f"Enter element [{r_idx}][{c_idx}]: ")))
    grid.append(row_buffer)

print("\nMatrix:")
for row_buffer in grid:
    print(" ".join(f"{element}" for element in row_buffer))

symmetric_flag = True

for r_idx in range(dim_size): # scan transpositions
    for c_idx in range(dim_size):
        if grid[r_idx][c_idx] != grid[c_idx][r_idx]:
            symmetric_flag = False
            break
    if not symmetric_flag:
        break

if symmetric_flag: # log outcome conditional
    print("\nThe matrix is a Symmetric Matrix.")
else:
    print("\nThe matrix is NOT a Symmetric Matrix.")
```

79. Write a program to Find row-wise sum.

```python
total_rows = int(input("Enter number of rows: "))
total_cols = int(input("Enter number of columns: "))

target_grid = []

print("\nEnter the elements of the matrix:")
for i in range(total_rows):
    sub_row = []
    for j in range(total_cols):
        sub_row.append(int(input(f"Enter element [{i}][{j}]: ")))
    target_grid.append(sub_row)

print("\nMatrix:")
for sub_row in target_grid:
    print(" ".join(f"{item}" for item in sub_row))

print("\nRow-wise Sum:")
for i in range(total_rows): # process items inside single row context
    running_row_total = 0
    for j in range(total_cols):
        running_row_total += target_grid[i][j]
    print(f"Sum of Row {i + 1} = {running_row_total}")
```

80. Write a program to Find column-wise sum.

```python
row_boundary = int(input("Enter number of rows: "))
col_boundary = int(input("Enter number of columns: "))

working_grid = []

print("\nEnter the elements of the matrix:")
for r in range(row_boundary):
    element_buffer = []
    for c in range(col_boundary):
        element_buffer.append(int(input(f"Enter element [{r}][{c}]: ")))
    working_grid.append(element_buffer)

print("\nMatrix:")
for element_buffer in working_grid:
    print(" ".join(f"{digit}" for digit in element_buffer))

print("\nColumn-wise Sum:")
for c in range(col_boundary): # process vertical columns sequentially
    running_col_total = 0
    for r in range(row_boundary):
        running_col_total += working_grid[r][c]
    print(f"Sum of Column {c + 1} = {running_col_total}")
```
