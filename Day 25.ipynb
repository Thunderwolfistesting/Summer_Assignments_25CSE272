97. Write a program to Merge two sorted arrays.

```python
def merge_arrays(left_part, right_part):
    combined = []
    i_idx, j_idx = 0, 0

    while i_idx < len(left_part) and j_idx < len(right_part):
        if left_part[i_idx] <= right_part[j_idx]:
            combined.append(left_part[i_idx])
            i_idx += 1
        else:
            combined.append(right_part[j_idx])
            j_idx += 1

    while i_idx < len(left_part):
        combined.append(left_part[i_idx])
        i_idx += 1

    while j_idx < len(right_part):
        combined.append(right_part[j_idx])
        j_idx += 1

    return combined

def alternate_merge_sort(sequence):
    if len(sequence) <= 1:
        return sequence
    
    midpoint = len(sequence) // 2
    left_slice = sequence[:midpoint]
    right_slice = sequence[midpoint:]
    
    sorted_left = alternate_merge_sort(left_slice)
    sorted_right = alternate_merge_sort(right_slice)

    return merge_arrays(sorted_left, sorted_right)

sample_nums = [44, 11, 22, 99, 10, 55, 33, 77, 66]
print(alternate_merge_sort(sample_nums))
```

98. Write a program to Find common characters in strings.

```python
first_input = input("Enter first string: ")
second_input = input("Enter second string: ")

print(f"your input: {first_input}")
print(f"your input: {second_input}", end=" ")
intersect_chars = ""

for char_a in first_input:
    is_found = False

    for char_b in second_input:
        if char_a == char_b:
            is_found = True
            break

    if is_found:
        is_duplicate = False

        for logged_char in intersect_chars:
            if logged_char == char_a:
                is_duplicate = True
                break

        if not is_duplicate:
            intersect_chars += char_a

print(f"Common characters: {intersect_chars}")
```

99. Write a program to Sort names alphabetically.

```python
total_names = int(input("Enter the number of names: "))

name_collection = []

for idx in range(total_names):
    received_name = input(f"Enter name {idx + 1}: ")
    name_collection.append(received_name)

for idx_i in range(total_names):
    for idx_j in range(0, total_names - idx_i - 1):
        if name_collection[idx_j] > name_collection[idx_j + 1]:
            temporary_holder = name_collection[idx_j]
            name_collection[idx_j] = name_collection[idx_j + 1]
            name_collection[idx_j + 1] = temporary_holder

print("Names in alphabetical order:")
for received_name in name_collection:
    print(received_name)
```
