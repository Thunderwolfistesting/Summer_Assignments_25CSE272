21. Write a program to Convert decimal to binary.

```python
number = int(input("Enter a number: "))
bin_str = ""

if number == 0:
    bin_str = "0"
else:
    while number > 0:
        bin_str = str(number % 2) + bin_str # append remainder
        number //= 2 # divide by 2

print(f"Binary number: {bin_str}")
```

22. Write a program to Convert binary to decimal.

```python
binary_input = input("Enter a binary number: ")
dec_value = 0
mult = 0

for bit in binary_input[::-1]: # reverse loop
    dec_value += int(bit) * (2 ** mult) # add power value
    mult += 1

print(f"Decimal number: {dec_value}")
```

23. Write a program to Count set bits in a number.

```python
val = int(input("Enter a number: "))
bit_count = 0

while val > 0:
    bit_count += val & 1 # check last bit
    val >>= 1 # shift right

print(f"Number of set bits: {bit_count}")
```

24. Write a program to Find x^n without pow().

```python
base = int(input("Enter the base (x): "))
exp = int(input("Enter the exponent (n): "))
ans = 1

for _ in range(exp): # multiply repeatedly
    ans *= base

print(f"{base} ^ {exp} = {ans}") # print final total
```
