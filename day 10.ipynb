37. Write a program to Print star pyramid.
```text
    *
   ***
  *****
 *******
*********
```

````python
limit = 5

for i in range(1, limit + 1):
    gap = limit - i # spacing calc
    count = 2 * i - 1 # odd count formula
    print(" " * gap + "*" * count)
````

38. Write a program to Print reverse pyramid.
```text
*********
 *******
  *****
   ***
    *
```

````python
height = 5

for i in range(height, 0, -1): # decrement down
    offset = height - i
    star_len = 2 * i - 1 # compute remaining width
    print(" " * offset + "*" * star_len)
````

39. Write a program to Print number pyramid.
```text
    1
   121
  12321
 1234321
123454321
```

````python
num_rows = 5

for i in range(1, num_rows + 1):
    print(" " * (num_rows - i), end="") # leading indent
    
    for j in range(1, i + 1):
        print(j, end="")
        
    for k in range(i - 1, 0, -1): # descending sequence
        print(k, end="")
        
    print()
````

40. Write a program to Print character pyramid.
```text
    A
   ABA
  ABCBA
 ABCDCBA
ABCDEDCBA
```

````python
char_rows = 5

for i in range(1, char_rows + 1):
    print(" " * (char_rows - i), end="") # apply gaps
    
    for j in range(i):
        print(chr(65 + j), end="")
        
    for k in range(i - 2, -1, -1): # count letters down
        print(chr(65 + k), end="")
        
    print()
````
