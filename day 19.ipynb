73. Write a program to Add matrices.

```python
mat_x = [[2, 4],
         [6, 8]]

mat_y = [[12, 12],
         [12, 12]]

total_mat = []

for r_idx in range(2):
    temp_row = []
    for c_idx in range(2):
        temp_row.append(mat_x[r_idx][c_idx] + mat_y[r_idx][c_idx]) # compute element totals
    total_mat.append(temp_row)

print("Sum of matrices:")
for r_idx in range(2):
    for c_idx in range(2):
        print(f"{total_mat[r_idx][c_idx]}", end=" ") # display inline spacing
    print()
```

74. Write a program to Subtract matrices.

```python
first_matrix = [[15, 20],
                [25, 30]]

second_matrix = [[5, 10],
                 [15, 20]]

diff_matrix = []

for row in range(2):
    accum_row = []
    for col in range(2):
        accum_row.append(first_matrix[row][col] - second_matrix[row][col]) # subtract corresponding cells
    diff_matrix.append(accum_row)

print("Subtraction of matrices:")
for row in range(2):
    for col in range(2):
        print(f"{diff_matrix[row][col]}", end=" ") # map grid visually
    print()
```

75. Write a program to Transpose matrix.

```python
rows_cnt = int(input("Enter number of rows: "))
cols_cnt = int(input("Enter number of columns: "))

base_matrix = []

print("Enter matrix elements:")
for i in range(rows_cnt):
    row_buffer = []
    for j in range(cols_cnt):
        val = int(input(f"Enter element A[{i}][{j}]: ")) # record nested values
        row_buffer.append(val)
    base_matrix.append(row_buffer)

print("\nInput Matrix:")
for i in range(rows_cnt):
    for j in range(cols_cnt):
        print(f"{base_matrix[i][j]}", end=" ")
    print()

print("\nTranspose of the Matrix:")
for i in range(cols_cnt):
    for j in range(rows_cnt):
        print(f"{base_matrix[j][i]}", end=" ") # swap dimension coordinates
    print()
```

76. Write a program to Find diagonal sum.

```python
grid_dimension = int(input("Enter the size of the square matrix: "))

target_matrix = []

print("Enter matrix elements:")
for r in range(grid_dimension):
    inner_collector = []
    for c in range(grid_dimension):
        element_input = int(input(f"Enter element A[{r}][{c}]: "))
        inner_collector.append(element_input)
    target_matrix.append(inner_collector) # load structural layer

print("\nInput Matrix:")
for r in range(grid_dimension):
    for c in range(grid_dimension):
        print(f"{target_matrix[r][c]}", end=" ")
    print()

trace_sum = 0

for idx in range(grid_dimension):
    trace_sum = trace_sum + target_matrix[idx][idx] # accumulate diagonal positions

print(f"\nDiagonal Sum = {trace_sum}")
```
