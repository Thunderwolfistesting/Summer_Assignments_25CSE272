89. Write a program to Find first non-repeating character.

```python
s = input("Enter a string: ")

print("Input: " + s)

i = 0

while i < len(s):
    count = 0
    j = 0

    while j < len(s):
        if s[i] == s[j]:
            count = count + 1
        j = j + 1

    if count == 1:
        print("First non-repeating character: " + s[i])
        break

    i = i + 1

if i == len(s):
    print("No non-repeating character found.")
```

90. Write a program to Find first repeating character.

```python
s = input("Enter a string: ")

print("Input: " + s)

i = 0

while i < len(s):
    j = i + 1

    while j < len(s):
        if s[i] == s[j]:
            print("First repeating character: " + s[i])
            break
        j = j + 1

    if j < len(s):
        break

    i = i + 1

if i == len(s):
    print("No repeating character found.")
```

91. Write a program to Check anagram strings.

```python
s1 = input("Enter first string: ")
s2 = input("Enter second string: ")

print("First String: " + s1)
print("Second String: " + s2)

if len(s1) != len(s2):
    print("Strings are not anagrams.")
else:
    used = ""
    i = 0
    flag = 1

    while i < len(s1):
        ch = s1[i]

        found = 0
        k = 0
        while k < len(used):
            if used[k] == ch:
                found = 1
                break
            k = k + 1

        if found == 0:
            c1 = 0
            c2 = 0

            j = 0
            while j < len(s1):
                if s1[j] == ch:
                    c1 = c1 + 1
                j = j + 1

            j = 0
            while j < len(s2):
                if s2[j] == ch:
                    c2 = c2 + 1
                j = j + 1

            if c1 != c2:
                flag = 0
                break

            used = used + ch

        i = i + 1

    if flag == 1:
        print("Strings are anagrams.")
    else:
        print("Strings are not anagrams.")
```

92. Write a program to Find maximum occurring character.

```python
s = input("Enter a string: ")

print("Input: " + s)

maxCount = 0
maxChar = ""

i = 0

while i < len(s):
    count = 0
    j = 0

    while j < len(s):
        if s[i] == s[j]:
            count = count + 1
        j = j + 1

    if count > maxCount:
        maxCount = count
        maxChar = s[i]

    i = i + 1

print("Maximum occurring character: " + maxChar)
print("Number of occurrences: " + str(maxCount))
```
