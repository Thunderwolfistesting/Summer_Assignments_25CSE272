9. Write a program to Check whether a number is prime.

```python
val = int(input("Enter value: "))
status = True

for i in range(2, int(val**0.5) + 1):
    if val % i == 0:
        status = False
        break

if status and val > 1:
    print(f"{val} is a prime number")
else:
    print(f"{val} is not a prime number")
```

10. Write a program to Print prime numbers in a range.

```python
lower = 2 # start limit
upper = 12 # end limit

for num in range(lower, upper + 1):
    is_prime = True
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            is_prime = False
            break
    if is_prime:
        print(num)
```

11. Write a program to Find GCD of two numbers.

```python
a = int(input("First number: "))
b = int(input("Second number: "))

result = 1
# count backwards to find gcd faster
for i in range(min(a, b), 0, -1):
    if a % i == 0 and b % i == 0:
        result = i
        break

print(f"The GCD of {a} and {b} is {result}")
```

12. Write a program to Find LCM of two numbers.

```python
x = int(input("First number: "))
y = int(input("Second number: "))

step = max(x, y)

while True:
    if step % x == 0 and step % y == 0:
        lcm_val = step
        break
    step += 1 # try next multiple

print(f"LCM of {x} and {y} is {lcm_val}")
```
