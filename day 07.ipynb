```python
# factorial using recursion
def fact(x):
    if x <= 1:
        return 1
    return x * fact(x - 1)

n = int(input("Number: "))

# check negative input
if n < 0:
    print("Invalid input")
else:
    print("Result =", fact(n))
```

26. Write a program to Recursive Fibonacci.

```python
# recursive fibonacci
def fib(n):
    if n < 2:
        return n
    return fib(n - 1) + fib(n - 2)

terms = int(input("How many terms? "))

# print sequence
print("Series:")
for j in range(terms):
    print(fib(j), end=" ")
```

27. Write a program to Recursive sum of digits.

```python
# digit sum function
def digit_sum(num):
    if num == 0:
        return 0
    return num % 10 + digit_sum(num // 10)

value = int(input("Enter value: "))

# display answer
print("Digit Sum:", digit_sum(abs(value)))
```

28. Write a program to Recursive reverse number.

```python
# reverse through recursion
def rev_num(n, ans=0):
    if n == 0:
        return ans
    return rev_num(n // 10, ans * 10 + n % 10)

number = int(input("Input number: "))

# print reversed value
print("Reverse:", rev_num(number))
```
