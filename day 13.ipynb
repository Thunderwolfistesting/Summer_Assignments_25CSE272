49. Write a program to Input and display array.

```python
num_elements = int(input("Enter number of elements: "))
data_list = []

for idx in range(num_elements): # populate items sequentially
    val = int(input(f"Enter element {idx + 1}: "))
    data_list.append(val)

print("Array elements are:")
for element in data_list: # dump items to stdout
    print(element, end=" ")
```

50. Write a program to Find sum and average of array.

```python
size_limit = int(input("Enter number of elements: "))
values = []

for _ in range(size_limit):
    values.append(int(input(f"Enter element {len(values) + 1}: ")))

grand_total = 0
for number in values: # manual aggregation step
    grand_total += number
    
avg_result = grand_total / size_limit # divide tracking length

print("Array:", values)
print("Sum =", grand_total)
print("Average =", avg_result)
```

51. Write a program to Find largest and smallest element.

```python
total_count = int(input("Enter number of elements: "))
numbers = []

for i in range(total_count):
    numbers.append(int(input(f"Enter element {i + 1}: ")))

print("Array elements are:")
for x in numbers: # list items horizontally
    print(x, end=" ")

max_val = numbers[0]
min_val = numbers[0]

for current in numbers:
    if current > max_val: # peak bounds validation
        max_val = current
    if current < min_val:
        min_val = current

print("")
print("Largest element =", max_val)
print("Smallest element =", min_val)
```

52. Write a program to Count even and odd elements.

```python
qty = int(input("Enter number of elements: "))
collection = []

for i in range(qty):
    collection.append(int(input(f"Enter element {i + 1}: ")))

print("Array elements are:")
for value in collection: # render flat representation
    print(value, end=" ")

evens = 0
odds = 0

for item in collection:
    if item % 2 == 0: # parity analysis
        evens += 1
    else:
        odds += 1

print("")
print("Number of even elements =", evens)
print("Number of odd elements =", odds)
```
