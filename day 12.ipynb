45. Write a program to Write function for palindrome.

```python
def check_palindrome(string_val):
    reversed_str = ""
    for char in string_val: # assemble backwards
        reversed_str = char + reversed_str
    return string_val == reversed_str

user_word = input("Enter a word: ")

if check_palindrome(user_word): # evaluate string comparison
    print(f'"{user_word}" is a palindrome.')
else:
    print(f'"{user_word}" is not a palindrome.')
```

46. Write a program to Write function for Armstrong.

```python
def verify_armstrong(value):
    text_num = str(value)
    power = len(text_num) # calculate length tracking
    grand_total = 0
    
    for digit in text_num:
        grand_total += int(digit) ** power
    return grand_total == value # matching validation

input_digit = int(input("Enter a number: "))

if verify_armstrong(input_digit):
    print(f"{input_digit} is an Armstrong number.")
else:
    print(f"{input_digit} is not an Armstrong number.")
```

47. Write a program to Write function for Fibonacci.

```python
def build_fibonacci(count_limit):
    first, second = 0, 1
    
    for _ in range(count_limit):
        print(first, end=" ") # display element sequence
        nxt = first + second
        first = second
        second = nxt # progress positions sequential

items_num = int(input("Enter the number of terms: "))
print("Fibonacci series:")
build_fibonacci(items_num)
```

48. Write a program to Write function for perfect number.

```python
def check_perfect_num(item):
    sum_divisors = 0
    
    for divisor in range(1, item):
        if item % divisor == 0: # factor discovery logic
            sum_divisors += divisor
            
    return sum_divisors == item # verify final status

given_val = int(input("Enter a number: "))

if check_perfect_num(given_val):
    print(f"{given_val} is a Perfect Number.")
else:
    print(f"{given_val} is not a Perfect Number.")
```
