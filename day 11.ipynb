41. Write a program to Write function to find sum of two numbers.

```python
def add_together(x, y):
    return x + y # return the addition

val1 = int(input("Enter first number: "))
val2 = int(input("Enter second number: "))

total = add_together(val1, val2) # run calculation function
print(f"Sum of {val1} and {val2} = {total}")
```

42. Write a program to Write function to find maximum.

```python
def get_highest(n1, n2):
    if n1 > n2: # check higher value
        return n1
    else:
        return n2

first_num = int(input("Enter first number: "))
second_num = int(input("Enter second number: "))

top_val = get_highest(first_num, second_num) # save returned value
print(f"The maximum number among {first_num} and {second_num} is {top_val}")
```

43. Write a program to Write function to check prime.

```python
def is_prime(target):
    if target <= 1:
        return False
    
    for check_val in range(2, target):
        if target % check_val == 0: # factoring test
            return False
            
    return True

test_element = int(input("Enter a number: "))

if is_prime(test_element): # execute conditional match
    print(f"{test_element} is a prime number")
else:
    print(f"{test_element} is not a prime number")
```

44. Write a program to Write function to find factorial.

```python
def compute_factorial(num):
    running_total = 1
    
    for step in range(1, num + 1): # loop limits
        running_total *= step
    return running_total # send result back

user_num = int(input("Enter a number: "))
outcome = compute_factorial(user_num)
print(f"The factorial of {user_num} is {outcome}")
```
