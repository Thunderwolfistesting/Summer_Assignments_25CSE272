117. Write a program to Create student record system using arrays and strings.

```python
student_names = []
student_marks = []

total_entries = int(input("Enter the number of students: "))

for idx in range(total_entries):
    input_name = input(f"Enter name of student {idx + 1}: ")
    input_mark = int(input(f"Enter marks of {input_name}: "))

    student_names.append(input_name)
    student_marks.append(input_mark)

print("\n----- Student Records -----")
for idx in range(len(student_names)):
    print(f"Student {idx + 1}: Name = {student_names[idx]}, Marks = {student_marks[idx]}")

mark_average = sum(student_marks) / len(student_marks)

print("\n----- Statistics -----")
print(f"Total Students : {len(student_names)}")
print(f"Highest Marks  : {max(student_marks)}")
print(f"Lowest Marks   : {min(student_marks)}")
print(f"Average Marks  : {mark_average:.2f}")

search_query = input("\nEnter student name to search: ")

if search_query in student_names:
    found_index = student_names.index(search_query)
    print(f"{search_query} scored {student_marks[found_index]} marks.")
else:
    print(f"Student '{search_query}' not found.")
```

118. Write a program to Create mini library system.

```python
inventory_list = []
book_count = int(input("Enter the number of books: "))

for idx in range(book_count):
    title_input = input(f"Enter book {idx + 1} name: ")
    inventory_list.append(title_input)

print("\n----- Library Books -----")
for idx in range(len(inventory_list)):
    print(f"{idx + 1}. {inventory_list[idx]}")

lookup_title = input("\nEnter book name to search: ")

if lookup_title in inventory_list:
    print(f'"{lookup_title}" is available in the library.')
else:
    print(f'"{lookup_title}" is not available.')

discard_title = input("\nEnter book name to remove: ")

if discard_title in inventory_list:
    inventory_list.remove(discard_title)
    print(f'"{discard_title}" has been removed.')
else:
    print(f'"{discard_title}" not found.')

print("\n----- Updated Library -----")
for idx in range(len(inventory_list)):
    print(f"{idx + 1}. {inventory_list[idx]}")

print(f"\nTotal books available: {len(inventory_list)}")
```

119. Write a program to Create mini employee management system.

```python
staff_names = []
staff_salaries = []

total_staff = int(input("Enter the number of employees: "))

for idx in range(total_staff):
    name_entry = input(f"Enter employee {idx + 1} name: ")
    salary_entry = float(input(f"Enter salary of {name_entry}: "))

    staff_names.append(name_entry)
    staff_salaries.append(salary_entry)

print("\n----- Employee Records -----")
for idx in range(len(staff_names)):
    print(f"Employee {idx + 1}: Name = {staff_names[idx]}, Salary = ${staff_salaries[idx]:.2f}")

lookup_staff = input("\nEnter employee name to search: ")

if lookup_staff in staff_names:
    match_idx = staff_names.index(lookup_staff)
    print(f"{lookup_staff}'s Salary = ${staff_salaries[match_idx]:.2f}")
else:
    print("Employee not found.")

print("\n----- Salary Statistics -----")
print(f"Total Employees : {len(staff_names)}")
print(f"Highest Salary  : ${max(staff_salaries):.2f}")
print(f"Lowest Salary   : ${min(staff_salaries):.2f}")
print(f"Average Salary  : ${sum(staff_salaries) / len(staff_salaries):.2f}")
```

120. Write a program to Develop complete mini project using arrays, strings and functions.

```python
global_names = []
global_marks = []

def register_new_student():
    input_name = input("Enter student name: ")
    input_mark = int(input("Enter marks: "))
    global_names.append(input_name)
    global_marks.append(input_mark)
    print(f"{input_name} added successfully.\n")

def print_all_students():
    if len(global_names) == 0:
        print("No student records found.\n")
    else:
        print("\n----- Student Records -----")
        for idx in range(len(global_names)):
            print(f"{idx + 1}. Name: {global_names[idx]}, Marks: {global_marks[idx]}")
        print()

def find_student_by_name():
    target_name = input("Enter student name to search: ")
    if target_name in global_names:
        found_idx = global_names.index(target_name)
        print(f"{target_name} scored {global_marks[found_idx]} marks.\n")
    else:
        print("Student not found.\n")

def compute_score_statistics():
    if len(global_marks) == 0:
        print("No records available.\n")
    else:
        print("\n----- Statistics -----")
        print(f"Total Students : {len(global_names)}")
        print(f"Highest Marks  : {max(global_marks)}")
        print(f"Lowest Marks   : {min(global_marks)}")
        print(f"Average Marks  : {sum(global_marks) / len(global_marks):.2f}\n")

while True:
    print("===== Student Management System =====")
    print("1. Add Student")
    print("2. Display Students")
    print("3. Search Student")
    print("4. Statistics")
    print("5. Exit")

    user_selection = input("Enter your choice: ")

    if user_selection == "1":
        register_new_student()
    elif user_selection == "2":
        print_all_students()
    elif user_selection == "3":
        find_student_by_name()
    elif user_selection == "4":
        compute_score_statistics()
    elif user_selection == "5":
        print("Thank you! Exiting program.")
        break
    else:
        print("Invalid choice. Try again.\n")
```
